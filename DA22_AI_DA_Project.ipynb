```python

# ============================================================
# MULTIVARIATE ANALYSIS - INSTAGRAM DATA ANALYSIS PROJECT
# ============================================================

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.graph_objects as go

from plotly.subplots import make_subplots
from IPython.display import display, Markdown

# ------------------------------------------------------------
# 1. DATA CLEANING (PRESERVE ORIGINAL DATAFRAMES)
# ------------------------------------------------------------

users = users_df.copy()
posts = posts_df.copy()
comments = comments_df.copy()

# Remove duplicate records
users = users.drop_duplicates()
posts = posts.drop_duplicates()
comments = comments.drop_duplicates()

# Remove duplicate primary keys
users = users.drop_duplicates(subset='User_ID', keep='first')
posts = posts.drop_duplicates(subset='Post_ID', keep='first')
comments = comments.drop_duplicates(subset='Comment_ID', keep='first')

# Standardize categorical columns
def clean_categories(df, columns):
    for col in columns:
        df[col] = (
            df[col].astype('string')
            .str.strip()
            .str.replace(r'\s+', ' ', regex=True)
            .str.title()
            .fillna('Unknown')
        )
    return df

users = clean_categories(
    users, ['Country', 'Account_Type']
)

posts = clean_categories(
    posts, ['Post_Type', 'Topic']
)

comments = clean_categories(
    comments, ['Sentiment']
)

# Handle missing numerical values
def fill_numeric(df, columns):
    for col in columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')
        df[col] = df[col].fillna(df[col].median())
    return df

users = fill_numeric(
    users, ['Followers_Count', 'Age']
)

posts = fill_numeric(
    posts, [
        'Likes', 'Comments_Count',
        'Shares', 'Saves', 'Reach', 'Impressions'
    ]
)

# Handle outliers using IQR capping
def cap_outliers(df, columns):
    for col in columns:
        q1 = df[col].quantile(0.25)
        q3 = df[col].quantile(0.75)
        iqr = q3 - q1

        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr

        df[col] = df[col].clip(lower, upper)

    return df

users = cap_outliers(
    users, ['Followers_Count', 'Age']
)

posts = cap_outliers(
    posts, [
        'Likes', 'Comments_Count',
        'Shares', 'Saves', 'Reach', 'Impressions'
    ]
)

# Prepare comments data
comments['Comment_Text'] = comments['Comment_Text'].fillna('')
comments['Like_Count'] = pd.to_numeric(
    comments['Like_Count'], errors='coerce'
).fillna(comments['Like_Count'].median())

comments['Reply_Count'] = pd.to_numeric(
    comments['Reply_Count'], errors='coerce'
).fillna(comments['Reply_Count'].median())


# ------------------------------------------------------------
# 2. MERGE USERS AND POSTS
# ------------------------------------------------------------

merged_df = posts.merge(
    users[['User_ID', 'Followers_Count', 'Country', 'Account_Type']],
    on='User_ID',
    how='left',
    validate='many_to_one'
)

# Calculate total engagement
merged_df['Engagement'] = (
    merged_df['Likes'] +
    merged_df['Comments_Count'] +
    merged_df['Shares'] +
    merged_df['Saves']
)


# ------------------------------------------------------------
# 3. GROUP PERFORMANCE ANALYSIS
# ------------------------------------------------------------

# Post Type vs Topic vs Engagement
post_performance = (
    merged_df.groupby(
        ['Post_Type', 'Topic'], as_index=False,
        observed=True
    )
    .agg(
        Avg_Engagement=('Engagement', 'mean'),
        Total_Engagement=('Engagement', 'sum'),
        Avg_Reach=('Reach', 'mean'),
        Avg_Likes=('Likes', 'mean')
    )
)

# Country vs Account Type vs User Count
country_performance = (
    users.groupby(
        ['Country', 'Account_Type'], as_index=False,
        observed=True
    )
    .agg(Users_Count=('User_ID', 'nunique'))
)

# Followers vs Reach vs Likes
correlation_data = merged_df[
    ['Followers_Count', 'Reach', 'Likes']
].corr(method='pearson')


# ------------------------------------------------------------
# 4. MATPLOTLIB + SEABORN MULTIVARIATE SUBPLOTS
# 2 ROWS x 3 COLUMNS
# ------------------------------------------------------------

sns.set_theme(style='whitegrid', context='notebook')

fig = plt.figure(figsize=(23, 15))

# 1. Grouped Bar Chart:
# Post Type vs Topic vs Engagement

ax1 = fig.add_subplot(2, 3, 1)

sns.barplot(
    data=post_performance,
    x='Post_Type',
    y='Avg_Engagement',
    hue='Topic',
    ax=ax1,
    errorbar=None,
    palette='viridis'
)

ax1.set_title(
    'Post Type vs Topic vs Engagement',
    fontsize=13, fontweight='bold'
)
ax1.set_xlabel('Post Type')
ax1.set_ylabel('Average Engagement')
ax1.tick_params(axis='x', rotation=35)
ax1.legend(title='Topic', fontsize=8, bbox_to_anchor=(1, 1))


# 2. Correlation Heatmap:
# Followers vs Reach vs Likes

ax2 = fig.add_subplot(2, 3, 2)

sns.heatmap(
    correlation_data,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    vmin=-1,
    vmax=1,
    ax=ax2
)

ax2.set_title(
    'Followers vs Reach vs Likes - Correlation',
    fontsize=13, fontweight='bold'
)


# 3. Grouped Bar Chart:
# Country vs Account Type vs Users Count

ax3 = fig.add_subplot(2, 3, 3)

sns.barplot(
    data=country_performance,
    x='Country',
    y='Users_Count',
    hue='Account_Type',
    ax=ax3,
    errorbar=None,
    palette='Set2'
)

ax3.set_title(
    'Country vs Account Type vs Users Count',
    fontsize=13, fontweight='bold'
)
ax3.set_xlabel('Country')
ax3.set_ylabel('Number of Users')
ax3.tick_params(axis='x', rotation=45)
ax3.legend(title='Account Type', fontsize=8)


# ------------------------------------------------------------
# 5. STATIC 3D SCATTER PLOTS
# ------------------------------------------------------------

# 4. Post Type vs Topic vs Engagement (3D)

ax4 = fig.add_subplot(2, 3, 4, projection='3d')

post_performance['Post_Code'] = pd.factorize(
    post_performance['Post_Type']
)[0]

post_performance['Topic_Code'] = pd.factorize(
    post_performance['Topic']
)[0]

sc4 = ax4.scatter(
    post_performance['Post_Code'],
    post_performance['Topic_Code'],
    post_performance['Avg_Engagement'],
    c=post_performance['Avg_Engagement'],
    cmap='viridis',
    s=65,
    alpha=0.8
)

ax4.set_title('Post Type vs Topic vs Engagement (3D)')
ax4.set_xlabel('Post Type')
ax4.set_ylabel('Topic')
ax4.set_zlabel('Average Engagement')

ax4.set_xticks(
    post_performance['Post_Code'].unique()
)
ax4.set_xticklabels(
    post_performance.drop_duplicates('Post_Code')['Post_Type'],
    rotation=30
)

ax4.set_yticks(
    post_performance['Topic_Code'].unique()
)
ax4.set_yticklabels(
    post_performance.drop_duplicates('Topic_Code')['Topic'],
    rotation=30
)


# 5. Followers vs Reach vs Likes (3D)

ax5 = fig.add_subplot(2, 3, 5, projection='3d')

sample_data = merged_df.sample(
    n=min(3000, len(merged_df)),
    random_state=42
)

sc5 = ax5.scatter(
    sample_data['Followers_Count'],
    sample_data['Reach'],
    sample_data['Likes'],
    c=sample_data['Likes'],
    cmap='plasma',
    s=12,
    alpha=0.6
)

ax5.set_title('Followers vs Reach vs Likes (3D)')
ax5.set_xlabel('Followers Count')
ax5.set_ylabel('Reach')
ax5.set_zlabel('Likes')


# 6. Country vs Account Type vs Users Count (3D)

ax6 = fig.add_subplot(2, 3, 6, projection='3d')

country_performance['Country_Code'] = pd.factorize(
    country_performance['Country']
)[0]

country_performance['Account_Code'] = pd.factorize(
    country_performance['Account_Type']
)[0]

sc6 = ax6.scatter(
    country_performance['Country_Code'],
    country_performance['Account_Code'],
    country_performance['Users_Count'],
    c=country_performance['Users_Count'],
    cmap='coolwarm',
    s=80,
    alpha=0.8
)

ax6.set_title('Country vs Account Type vs Users Count (3D)')
ax6.set_xlabel('Country')
ax6.set_ylabel('Account Type')
ax6.set_zlabel('Users Count')

ax6.set_xticks(country_performance['Country_Code'].unique())
ax6.set_xticklabels(
    country_performance.drop_duplicates('Country_Code')['Country'],
    rotation=45
)

ax6.set_yticks(country_performance['Account_Code'].unique())
ax6.set_yticklabels(
    country_performance.drop_duplicates('Account_Code')['Account_Type']
)

plt.suptitle(
    'INSTAGRAM - MULTIVARIATE DATA ANALYSIS',
    fontsize=20,
    fontweight='bold',
    y=1.02
)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 6. INTERACTIVE PLOTLY 3D VISUALIZATIONS
# ------------------------------------------------------------

fig3d = make_subplots(
    rows=1,
    cols=3,
    specs=[[
        {'type': 'scene'},
        {'type': 'scene'},
        {'type': 'scene'}
    ]],
    subplot_titles=[
        'Post Type vs Topic vs Engagement',
        'Followers vs Reach vs Likes',
        'Country vs Account Type vs Users Count'
    ]
)

# Interactive 3D - Chart 1

fig3d.add_trace(
    go.Scatter3d(
        x=post_performance['Post_Type'],
        y=post_performance['Topic'],
        z=post_performance['Avg_Engagement'],
        mode='markers',
        marker=dict(
            size=6,
            color=post_performance['Avg_Engagement'],
            colorscale='Viridis',
            showscale=True,
            colorbar=dict(title='Engagement', x=0.28)
        ),
        text=post_performance['Total_Engagement'],
        hovertemplate=(
            'Post Type: %{x}<br>'
            'Topic: %{y}<br>'
            'Average Engagement: %{z:.2f}<br>'
            'Total Engagement: %{text}<extra></extra>'
        ),
        name='Post Engagement'
    ),
    row=1, col=1
)

# Interactive 3D - Chart 2

fig3d.add_trace(
    go.Scatter3d(
        x=sample_data['Followers_Count'],
        y=sample_data['Reach'],
        z=sample_data['Likes'],
        mode='markers',
        marker=dict(
            size=3,
            color=sample_data['Likes'],
            colorscale='Plasma',
            opacity=0.7,
            showscale=True,
            colorbar=dict(title='Likes', x=0.64)
        ),
        name='Followers Reach Likes',
        hovertemplate=(
            'Followers: %{x}<br>'
            'Reach: %{y}<br>'
            'Likes: %{z}<extra></extra>'
        )
    ),
    row=1, col=2
)

# Interactive 3D - Chart 3

fig3d.add_trace(
    go.Scatter3d(
        x=country_performance['Country'],
        y=country_performance['Account_Type'],
        z=country_performance['Users_Count'],
        mode='markers',
        marker=dict(
            size=7,
            color=country_performance['Users_Count'],
            colorscale='Cividis',
            showscale=True,
            colorbar=dict(title='Users', x=1.0)
        ),
        name='Country Account Users',
        hovertemplate=(
            'Country: %{x}<br>'
            'Account Type: %{y}<br>'
            'Users: %{z}<extra></extra>'
        )
    ),
    row=1, col=3
)

fig3d.update_layout(
    title=dict(
        text='Instagram - Interactive 3D Multivariate Analysis',
        x=0.5
    ),
    height=750,
    width=1800,
    template='plotly_white'
)

fig3d.update_scenes(
    xaxis_title='Post Type',
    yaxis_title='Topic',
    zaxis_title='Average Engagement',
    row=1, col=1
)

fig3d.update_scenes(
    xaxis_title='Followers Count',
    yaxis_title='Reach',
    zaxis_title='Likes',
    row=1, col=2
)

fig3d.update_scenes(
    xaxis_title='Country',
    yaxis_title='Account Type',
    zaxis_title='Users Count',
    row=1, col=3
)

fig3d.show()


# ------------------------------------------------------------
# 7. AUTOMATED KEY FINDINGS AND BUSINESS INSIGHTS
# ------------------------------------------------------------

# Best performing Post Type and Topic
best_post = post_performance.loc[
    post_performance['Avg_Engagement'].idxmax()
]

# Highest user-count Country and Account Type combination
best_country = country_performance.loc[
    country_performance['Users_Count'].idxmax()
]

# Correlations
corr_followers_reach = correlation_data.loc[
    'Followers_Count', 'Reach'
]

corr_followers_likes = correlation_data.loc[
    'Followers_Count', 'Likes'
]

corr_reach_likes = correlation_data.loc[
    'Reach', 'Likes'
]

# Display business insights
display(Markdown("## Automated Key Findings & Business Insights"))

display(Markdown(
    f"""
### 1. Post Performance
- Highest average engagement: **{best_post['Post_Type']}**
- Best performing topic: **{best_post['Topic']}**
- Average engagement: **{best_post['Avg_Engagement']:.2f}**

**Business Insight:** Prioritize content types and topics
with higher average engagement. Test similar content to
understand whether the performance can be repeated.

### 2. Followers, Reach and Likes
- Followers vs Reach correlation: **{corr_followers_reach:.3f}**
- Followers vs Likes correlation: **{corr_followers_likes:.3f}**
- Reach vs Likes correlation: **{corr_reach_likes:.3f}**

**Business Insight:** Study the strength of these relationships
to understand whether audience size and content exposure
are associated with higher likes.

### 3. Country and Account Type
- Highest user-count country: **{best_country['Country']}**
- Account type: **{best_country['Account_Type']}**
- User count: **{best_country['Users_Count']}**

**Business Insight:** Consider audience-specific content
strategies for countries and account types with larger
user populations.
"""
))

# Additional actionable recommendations
display(Markdown(
    """
### Actionable Recommendations

1. Focus on post types and topics with consistently high engagement.
2. Investigate posts with high reach but relatively low likes.
3. Compare follower size with post-level performance.
4. Develop country-specific content strategies.
5. Compare account types to identify the largest audience segments.
6. Monitor engagement metrics regularly to measure improvements.
"""
))
```

```python
# Find the  columnwise outliers in all the 3 files

import pandas as pd
import numpy as np

def find_outliers(df, name):
    print(f"\n{'='*50}")
    print(f"OUTLIER ANALYSIS: {name}")
    print(f"{'='*50}")

    # Select numerical columns and exclude ID columns
    num_cols = [
        col for col in df.select_dtypes(include=np.number).columns
        if 'id' not in col.lower()
    ]

    results = []

    for col in num_cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        outliers = df[(df[col] < lower) | (df[col] > upper)][col]

        results.append({
            'Column': col,
            'Q1': Q1,
            'Q3': Q3,
            'IQR': IQR,
            'Lower_Bound': lower,
            'Upper_Bound': upper,
            'Outlier_Count': len(outliers),
            'Outlier_Percentage': round(len(outliers) / df[col].notna().sum() * 100, 2),
            'Outlier_Values': outliers.tolist()
        })

    return pd.DataFrame(results)


# Find outliers in all three DataFrames
users_outliers = find_outliers(users_df, 'Users Data')
posts_outliers = find_outliers(posts_df, 'Posts Data')
comments_outliers = find_outliers(comments_df, 'Comments Data')


# Display results
display(users_outliers)
display(posts_outliers)
display(comments_outliers)
```

```python
import pandas as pd
import numpy as np

# Store all three DataFrames
datasets = {
    'Users': users_df,
    'Posts': posts_df,
    'Comments': comments_df
}

# Store outlier results
outlier_results = []

# Analyse each DataFrame
for dataset_name, df in datasets.items():

    # Select numerical columns and exclude ID columns
    numerical_cols = [
        col for col in df.select_dtypes(include=np.number).columns
        if 'id' not in col.lower()
    ]

    # Analyse each numerical column
    for col in numerical_cols:

        data = df[col].dropna()

        # Skip columns without valid numerical values
        if data.empty:
            continue

        # Calculate quartiles and IQR
        Q1 = data.quantile(0.25)
        Q3 = data.quantile(0.75)
        IQR = Q3 - Q1

        # Calculate lower and upper limits
        lower_limit = Q1 - (1.5 * IQR)
        upper_limit = Q3 + (1.5 * IQR)

        # Identify actual outlier values
        outliers = data[(data < lower_limit) | (data > upper_limit)]

        # Calculate outlier count and percentage
        outlier_count = len(outliers)
        outlier_percentage = (outlier_count / len(data)) * 100

        # Store results
        outlier_results.append({
            'Dataset': dataset_name,
            'Column': col,
            'Q1': round(Q1, 2),
            'Q3': round(Q3, 2),
            'IQR': round(IQR, 2),
            'Lower_Limit': round(lower_limit, 2),
            'Upper_Limit': round(upper_limit, 2),
            'Outlier_Count': outlier_count,
            'Outlier_Percentage': round(outlier_percentage, 2),
            'Outlier_Values': outliers.tolist()
        })

# Create a single summary DataFrame
outlier_summary = pd.DataFrame(outlier_results)

# Display complete results
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)

display(outlier_summary)
```

```python
# Find all the inconsistency in all the 3 files together

import pandas as pd
import numpy as np

# Store all three DataFrames
datasets = {
    'Users': users_df,
    'Posts': posts_df,
    'Comments': comments_df
}

# Store inconsistency results
inconsistency_results = []

# Analyse each dataset
for dataset_name, df in datasets.items():

    for col in df.columns:

        # Ignore missing values during inconsistency checks
        data = df[col].dropna()

        # Skip empty columns
        if data.empty:
            continue

        # 1. Check string columns
        if pd.api.types.is_object_dtype(df[col]) or pd.api.types.is_string_dtype(df[col]):

            # Leading and trailing spaces
            spaces = data[data.astype(str) != data.astype(str).str.strip()]

            if len(spaces) > 0:
                inconsistency_results.append({
                    'Dataset': dataset_name,
                    'Column': col,
                    'Issue': 'Leading/Trailing Spaces',
                    'Count': len(spaces),
                    'Values': spaces.unique().tolist()
                })

            # Blank strings
            blanks = data[data.astype(str).str.strip() == '']

            if len(blanks) > 0:
                inconsistency_results.append({
                    'Dataset': dataset_name,
                    'Column': col,
                    'Issue': 'Blank Values',
                    'Count': len(blanks),
                    'Values': blanks.unique().tolist()
                })

            # Inconsistent uppercase/lowercase or spacing variations
            normalized = data.astype(str).str.strip().str.lower()
            variants = pd.DataFrame({
                'Original': data.astype(str),
                'Normalized': normalized
            })

            inconsistent = variants.groupby('Normalized')['Original'].agg(
                lambda x: x.unique().tolist()
            )

            inconsistent = inconsistent[inconsistent.apply(len) > 1]

            if not inconsistent.empty:
                inconsistency_results.append({
                    'Dataset': dataset_name,
                    'Column': col,
                    'Issue': 'Case/Formatting Variations',
                    'Count': sum(inconsistent.apply(len)),
                    'Values': inconsistent.tolist()
                })

            # Mixed data types
            types = data.apply(lambda x: type(x).__name__).unique()

            if len(types) > 1:
                inconsistency_results.append({
                    'Dataset': dataset_name,
                    'Column': col,
                    'Issue': 'Mixed Data Types',
                    'Count': len(data),
                    'Values': types.tolist()
                })

        # 2. Check date columns
        if 'date' in col.lower():
            dates = pd.to_datetime(data, errors='coerce')
            invalid_dates = data[dates.isna()]

            if len(invalid_dates) > 0:
                inconsistency_results.append({
                    'Dataset': dataset_name,
                    'Column': col,
                    'Issue': 'Invalid Date Values',
                    'Count': len(invalid_dates),
                    'Values': invalid_dates.unique().tolist()
                })

        # 3. Check negative values in age/count columns
        if pd.api.types.is_numeric_dtype(df[col]):

            if any(word in col.lower() for word in
                   ['age', 'count', 'likes', 'shares', 'saves',
                    'reach', 'impressions', 'followers', 'following',
                    'comments', 'replies']):

                invalid = data[data < 0]

                if len(invalid) > 0:
                    inconsistency_results.append({
                        'Dataset': dataset_name,
                        'Column': col,
                        'Issue': 'Negative Values',
                        'Count': len(invalid),
                        'Values': invalid.unique().tolist()
                    })

            # Invalid age values
            if col.lower() == 'age':
                invalid_age = data[data > 120]

                if len(invalid_age) > 0:
                    inconsistency_results.append({
                        'Dataset': dataset_name,
                        'Column': col,
                        'Issue': 'Age Above 120',
                        'Count': len(invalid_age),
                        'Values': invalid_age.unique().tolist()
                    })


# Create one summary DataFrame
inconsistency_summary = pd.DataFrame(inconsistency_results)

# Display all inconsistencies
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)

display(inconsistency_summary)
```

```python

import pandas as pd
import numpy as np

# ----------------------------------
# 1. Create copies of the DataFrames
# ----------------------------------
users = users_df.copy()
posts = posts_df.copy()
comments = comments_df.copy()

# ----------------------------------
# 2. Handle missing values
# ----------------------------------

# Users
if 'Age' in users.columns:
    users['Age'] = pd.to_numeric(users['Age'], errors='coerce')
    users['Age'] = users['Age'].fillna(users['Age'].median())

if 'City' in users.columns:
    users['City'] = users['City'].fillna('Unknown')

# Posts
if 'Caption' in posts.columns:
    posts['Caption'] = posts['Caption'].fillna('Not provided')

# Comments
if 'Comment_Text' in comments.columns:
    comments['Comment_Text'] = comments['Comment_Text'].fillna('Not provided')

if 'Sentiment' in comments.columns:
    comments['Sentiment'] = comments['Sentiment'].fillna('Unknown')

# ----------------------------------
# 3. Convert numeric columns
# ----------------------------------
post_numeric = [
    'Likes', 'Comments_Count', 'Shares',
    'Saves', 'Reach', 'Impressions'
]

for col in post_numeric:
    if col in posts.columns:
        posts[col] = pd.to_numeric(posts[col], errors='coerce').fillna(0)

comment_numeric = ['Like_Count', 'Reply_Count']

for col in comment_numeric:
    if col in comments.columns:
        comments[col] = pd.to_numeric(
            comments[col], errors='coerce'
        ).fillna(0)

# ----------------------------------
# 4. Calculate engagement metrics
# ----------------------------------
posts['Engagement'] = (
    posts['Likes']
    + posts['Comments_Count']
    + posts['Shares']
    + posts['Saves']
)

posts['Engagement_Rate'] = np.where(
    posts['Reach'] > 0,
    posts['Engagement'] / posts['Reach'] * 100,
    0
)

# Convert post date
posts['Post_Date'] = pd.to_datetime(
    posts['Post_Date'], errors='coerce'
)

posts['Posting_Day'] = posts['Post_Date'].dt.day_name()

# ----------------------------------
# 5. Insight 1: Best-performing post types
# ----------------------------------
post_type = posts.groupby('Post_Type').agg(
    Total_Posts=('Post_ID', 'count'),
    Avg_Likes=('Likes', 'mean'),
    Avg_Reach=('Reach', 'mean'),
    Avg_Shares=('Shares', 'mean'),
    Avg_Saves=('Saves', 'mean'),
    Avg_Engagement_Rate=('Engagement_Rate', 'mean')
).reset_index()

post_type = post_type.sort_values(
    'Avg_Engagement_Rate', ascending=False
)

# ----------------------------------
# 6. Insight 2: Popular content topics
# ----------------------------------
topics = posts.groupby('Topic').agg(
    Total_Posts=('Post_ID', 'count'),
    Avg_Likes=('Likes', 'mean'),
    Avg_Reach=('Reach', 'mean'),
    Avg_Shares=('Shares', 'mean'),
    Avg_Saves=('Saves', 'mean'),
    Avg_Engagement_Rate=('Engagement_Rate', 'mean')
).reset_index()

topics = topics.sort_values(
    'Avg_Engagement_Rate', ascending=False
)

# ----------------------------------
# 7. Insight 3: Best posting days
# ----------------------------------
posting_days = posts.groupby('Posting_Day').agg(
    Total_Posts=('Post_ID', 'count'),
    Avg_Likes=('Likes', 'mean'),
    Avg_Reach=('Reach', 'mean'),
    Avg_Engagement_Rate=('Engagement_Rate', 'mean')
).reset_index()

day_order = [
    'Monday', 'Tuesday', 'Wednesday',
    'Thursday', 'Friday', 'Saturday', 'Sunday'
]

posting_days['Posting_Day'] = pd.Categorical(
    posting_days['Posting_Day'],
    categories=day_order,
    ordered=True
)

posting_days = posting_days.sort_values('Posting_Day')

# ----------------------------------
# 8. Insight 4: Target audience
# ----------------------------------
user_posts = posts.merge(
    users[['User_ID', 'City', 'Account_Type']],
    on='User_ID',
    how='left'
)

city_analysis = user_posts.groupby('City').agg(
    Total_Posts=('Post_ID', 'count'),
    Avg_Likes=('Likes', 'mean'),
    Avg_Reach=('Reach', 'mean'),
    Avg_Engagement_Rate=('Engagement_Rate', 'mean')
).reset_index()

city_analysis = city_analysis.sort_values(
    'Avg_Reach', ascending=False
)

account_analysis = user_posts.groupby('Account_Type').agg(
    Total_Posts=('Post_ID', 'count'),
    Avg_Likes=('Likes', 'mean'),
    Avg_Reach=('Reach', 'mean'),
    Avg_Engagement_Rate=('Engagement_Rate', 'mean')
).reset_index()

account_analysis = account_analysis.sort_values(
    'Avg_Engagement_Rate', ascending=False
)

# ----------------------------------
# 9. Insight 5: Audience feedback
# ----------------------------------
if 'Sentiment' in comments.columns:
    sentiment_analysis = comments.groupby('Sentiment').agg(
        Total_Comments=('Comment_ID', 'count'),
        Avg_Comment_Likes=('Like_Count', 'mean'),
        Avg_Replies=('Reply_Count', 'mean'),
        Total_Comment_Likes=('Like_Count', 'sum'),
        Total_Replies=('Reply_Count', 'sum')
    ).reset_index()
else:
    sentiment_analysis = pd.DataFrame()

# Overall comment interactions
comment_summary = pd.DataFrame({
    'Metric': [
        'Total Comments',
        'Average Comment Likes',
        'Average Comment Replies',
        'Total Comment Likes',
        'Total Comment Replies'
    ],
    'Value': [
        len(comments),
        comments['Like_Count'].mean(),
        comments['Reply_Count'].mean(),
        comments['Like_Count'].sum(),
        comments['Reply_Count'].sum()
    ]
})

# ----------------------------------
# 10. Overall Instagram KPIs
# ----------------------------------
overall_kpis = pd.DataFrame({
    'Metric': [
        'Total Posts',
        'Average Likes',
        'Average Reach',
        'Average Shares',
        'Average Saves',
        'Average Engagement Rate',
        'Average Post Comments'
    ],
    'Value': [
        len(posts),
        posts['Likes'].mean(),
        posts['Reach'].mean(),
        posts['Shares'].mean(),
        posts['Saves'].mean(),
        posts['Engagement_Rate'].mean(),
        posts['Comments_Count'].mean()
    ]
})

# ----------------------------------
# 11. Display all results
# ----------------------------------
tables = {
    'Overall Instagram KPIs': overall_kpis,
    'Best Performing Post Types': post_type,
    'Popular Content Topics': topics,
    'Best Posting Days': posting_days,
    'Top Cities': city_analysis.head(10),
    'Account Type Analysis': account_analysis,
    'Comment Sentiment Analysis': sentiment_analysis,
    'Comment Interactions': comment_summary
}

for title, table in tables.items():
    print('\n' + '=' * 65)
    print(title.upper())
    print('=' * 65)
    print(table.round(2).to_string(index=False))
```

```python

import pandas as pd
import numpy as np

# =====================================================
# 1. LOAD DATASETS
# =====================================================

users_df = pd.read_csv("users.csv")
posts_df = pd.read_csv("posts.csv")
comments_df = pd.read_csv("comments.csv")

users = users_df.copy()
posts = posts_df.copy()
comments = comments_df.copy()

# =====================================================
# 2. DATA CLEANING
# =====================================================

# Remove extra spaces from column text values
for df in [users, posts, comments]:
    for col in df.select_dtypes(include="object").columns:
        df[col] = df[col].str.strip()

# Handle missing user values
if "Age" in users.columns:
    users["Age"] = pd.to_numeric(users["Age"], errors="coerce")
    users["Age"] = users["Age"].fillna(users["Age"].median())

users["City"] = users["City"].fillna("Unknown")
users["Account_Type"] = users["Account_Type"].fillna("Unknown")

# Handle missing post values
posts["Caption"] = posts["Caption"].fillna("Not provided")
posts["Topic"] = posts["Topic"].fillna("Unknown")
posts["Post_Type"] = posts["Post_Type"].fillna("Unknown")

# Handle missing comment values
comments["Comment_Text"] = comments["Comment_Text"].fillna("Not provided")
comments["Sentiment"] = comments["Sentiment"].fillna("Unknown")

# Standardize category names
for col in ["City", "Account_Type"]:
    users[col] = users[col].astype(str).str.strip().str.title()

for col in ["Post_Type", "Topic"]:
    posts[col] = posts[col].astype(str).str.strip().str.title()

comments["Sentiment"] = (
    comments["Sentiment"].astype(str).str.strip().str.title()
)

# Convert numeric columns
post_numeric = [
    "Likes", "Comments_Count", "Shares",
    "Saves", "Reach", "Impressions"
]

for col in post_numeric:
    posts[col] = pd.to_numeric(posts[col], errors="coerce").fillna(0)

for col in ["Like_Count", "Reply_Count"]:
    comments[col] = pd.to_numeric(
        comments[col], errors="coerce"
    ).fillna(0)

# Convert dates
posts["Post_Date"] = pd.to_datetime(
    posts["Post_Date"], errors="coerce"
)

comments["Comment_Date"] = pd.to_datetime(
    comments["Comment_Date"], errors="coerce"
)

# =====================================================
# 3. CALCULATE ENGAGEMENT METRICS
# =====================================================

posts["Engagement"] = (
    posts["Likes"]
    + posts["Comments_Count"]
    + posts["Shares"]
    + posts["Saves"]
)

posts["Engagement_Rate"] = np.where(
    posts["Reach"] > 0,
    posts["Engagement"] / posts["Reach"] * 100,
    np.nan
)

posts["Posting_Day"] = posts["Post_Date"].dt.day_name()

# =====================================================
# 4. OVERALL INSTAGRAM KPIs
# =====================================================

overall_kpis = pd.DataFrame({
    "Metric": [
        "Total Users",
        "Total Posts",
        "Total Comments",
        "Average Likes",
        "Average Reach",
        "Average Shares",
        "Average Saves",
        "Average Engagement Rate (%)",
        "Average Post Comments",
        "Total Engagement"
    ],
    "Value": [
        users["User_ID"].nunique(),
        len(posts),
        len(comments),
        posts["Likes"].mean(),
        posts["Reach"].mean(),
        posts["Shares"].mean(),
        posts["Saves"].mean(),
        posts["Engagement_Rate"].mean(),
        posts["Comments_Count"].mean(),
        posts["Engagement"].sum()
    ]
})

# =====================================================
# 5. INSIGHT 1: BEST-PERFORMING POST TYPES
# =====================================================

post_type_analysis = posts.groupby("Post_Type").agg(
    Total_Posts=("Post_ID", "count"),
    Average_Likes=("Likes", "mean"),
    Average_Reach=("Reach", "mean"),
    Average_Shares=("Shares", "mean"),
    Average_Saves=("Saves", "mean"),
    Average_Engagement_Rate=("Engagement_Rate", "mean")
).reset_index()

post_type_analysis = post_type_analysis.sort_values(
    "Average_Engagement_Rate", ascending=False
)

# =====================================================
# 6. INSIGHT 2: POPULAR CONTENT TOPICS
# =====================================================

topic_analysis = posts.groupby("Topic").agg(
    Total_Posts=("Post_ID", "count"),
    Average_Likes=("Likes", "mean"),
    Average_Reach=("Reach", "mean"),
    Average_Shares=("Shares", "mean"),
    Average_Saves=("Saves", "mean"),
    Average_Engagement_Rate=("Engagement_Rate", "mean")
).reset_index()

topic_analysis = topic_analysis.sort_values(
    "Average_Engagement_Rate", ascending=False
)

# =====================================================
# 7. INSIGHT 3: BEST POSTING DAYS
# =====================================================

day_analysis = posts.groupby("Posting_Day").agg(
    Total_Posts=("Post_ID", "count"),
    Average_Likes=("Likes", "mean"),
    Average_Reach=("Reach", "mean"),
    Average_Engagement_Rate=("Engagement_Rate", "mean")
).reset_index()

day_order = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

day_analysis["Posting_Day"] = pd.Categorical(
    day_analysis["Posting_Day"],
    categories=day_order,
    ordered=True
)

day_analysis = day_analysis.sort_values("Posting_Day")

# =====================================================
# 8. INSIGHT 4: TARGET AUDIENCE - CITIES
# =====================================================

user_posts = posts.merge(
    users[["User_ID", "City", "Account_Type"]],
    on="User_ID",
    how="left"
)

city_analysis = user_posts.groupby("City").agg(
    Total_Posts=("Post_ID", "count"),
    Unique_Users=("User_ID", "nunique"),
    Average_Likes=("Likes", "mean"),
    Average_Reach=("Reach", "mean"),
    Average_Engagement_Rate=("Engagement_Rate", "mean")
).reset_index()

city_analysis = city_analysis.sort_values(
    "Average_Engagement_Rate", ascending=False
)

# =====================================================
# 9. TARGET AUDIENCE - ACCOUNT TYPES
# =====================================================

account_analysis = user_posts.groupby("Account_Type").agg(
    Total_Posts=("Post_ID", "count"),
    Unique_Users=("User_ID", "nunique"),
    Average_Likes=("Likes", "mean"),
    Average_Reach=("Reach", "mean"),
    Average_Shares=("Shares", "mean"),
    Average_Saves=("Saves", "mean"),
    Average_Engagement_Rate=("Engagement_Rate", "mean")
).reset_index()

account_analysis = account_analysis.sort_values(
    "Average_Engagement_Rate", ascending=False
)

# =====================================================
# 10. INSIGHT 5: AUDIENCE FEEDBACK AND SENTIMENT
# =====================================================

sentiment_analysis = comments.groupby("Sentiment").agg(
    Total_Comments=("Comment_ID", "count"),
    Average_Comment_Likes=("Like_Count", "mean"),
    Average_Replies=("Reply_Count", "mean"),
    Total_Comment_Likes=("Like_Count", "sum"),
    Total_Replies=("Reply_Count", "sum")
).reset_index()

sentiment_analysis["Comment_Percentage"] = (
    sentiment_analysis["Total_Comments"]
    / len(comments) * 100
)

# Overall comment interactions
comment_interactions = pd.DataFrame({
    "Metric": [
        "Total Comment Likes",
        "Total Comment Replies",
        "Average Comment Likes",
        "Average Comment Replies"
    ],
    "Value": [
        comments["Like_Count"].sum(),
        comments["Reply_Count"].sum(),
        comments["Like_Count"].mean(),
        comments["Reply_Count"].mean()
    ]
})

# =====================================================
# 11. DATA QUALITY CHECK
# =====================================================

data_quality = pd.DataFrame({
    "Dataset": ["Users", "Posts", "Comments"],
    "Total_Rows": [
        len(users), len(posts), len(comments)
    ],
    "Missing_Cells": [
        users.isnull().sum().sum(),
        posts.isnull().sum().sum(),
        comments.isnull().sum().sum()
    ],
    "Duplicate_Rows": [
        users.duplicated().sum(),
        posts.duplicated().sum(),
        comments.duplicated().sum()
    ]
})

# =====================================================
# 12. DISPLAY ALL RESULT TABLES
# =====================================================

tables = {
    "OVERALL INSTAGRAM KPIs": overall_kpis,
    "BEST PERFORMING POST TYPES": post_type_analysis,
    "POPULAR CONTENT TOPICS": topic_analysis,
    "BEST POSTING DAYS": day_analysis,
    "CITY ANALYSIS": city_analysis,
    "ACCOUNT TYPE ANALYSIS": account_analysis,
    "COMMENT SENTIMENT ANALYSIS": sentiment_analysis,
    "COMMENT INTERACTIONS": comment_interactions,
    "DATA QUALITY REPORT": data_quality
}

for title, table in tables.items():
    print("\n" + "=" * 75)
    print(title)
    print("=" * 75)
    print(table.round(2).to_string(index=False))

# =====================================================
# 13. AUTOMATICALLY PRINT 5 KEY INSIGHTS
# =====================================================

print("\n" + "=" * 75)
print("5 IMPORTANT INSTAGRAM GROWTH INSIGHTS")
print("=" * 75)

if not post_type_analysis.empty:
    best = post_type_analysis.iloc[0]
    print(f"\n1. BEST POST TYPE: {best['Post_Type']}")
    print(f"   Average engagement rate: {best['Average_Engagement_Rate']:.2f}%")
    print(f"   Average reach: {best['Average_Reach']:.2f}")
    print(f"   Average likes: {best['Average_Likes']:.2f}")
    print("   ACTION: Publish more content in this format.")

if not topic_analysis.empty:
    best = topic_analysis.iloc[0]
    print(f"\n2. TOP CONTENT TOPIC: {best['Topic']}")
    print(f"   Average engagement rate: {best['Average_Engagement_Rate']:.2f}%")
    print(f"   Average reach: {best['Average_Reach']:.2f}")
    print(f"   Average likes: {best['Average_Likes']:.2f}")
    print("   ACTION: Create more content around this topic.")

if not day_analysis.dropna(subset=["Average_Engagement_Rate"]).empty:
    best = day_analysis.loc[
        day_analysis["Average_Engagement_Rate"].idxmax()
    ]
    print(f"\n3. BEST POSTING DAY: {best['Posting_Day']}")
    print(f"   Average engagement rate: {best['Average_Engagement_Rate']:.2f}%")
    print(f"   Average reach: {best['Average_Reach']:.2f}")
    print("   ACTION: Test more posts on this day.")

if not city_analysis.empty:
    best = city_analysis.iloc[0]
    print(f"\n4. TOP CITY BY ENGAGEMENT: {best['City']}")
    print(f"   Unique users: {best['Unique_Users']}")
    print(f"   Average reach: {best['Average_Reach']:.2f}")
    print(f"   Average engagement rate: {best['Average_Engagement_Rate']:.2f}%")
    print("   ACTION: Test location-specific content for this city.")

if not account_analysis.empty:
    best = account_analysis.iloc[0]
    print(f"\n   TOP ACCOUNT TYPE: {best['Account_Type']}")
    print(f"   Average engagement rate: {best['Average_Engagement_Rate']:.2f}%")
    print(f"   Average reach: {best['Average_Reach']:.2f}")
    print("   ACTION: Study and target this audience segment.")

if not sentiment_analysis.empty:
    best = sentiment_analysis.loc[
        sentiment_analysis["Total_Comments"].idxmax()
    ]
    print(f"\n5. MOST COMMON SENTIMENT: {best['Sentiment']}")
    print(f"   Comments: {best['Total_Comments']}")
    print(f"   Comment share: {best['Comment_Percentage']:.2f}%")
    print("   ACTION: Respond to feedback and encourage conversations.")

# =====================================================
# 14. WHAT DOES NOT MATTER AS MUCH
# =====================================================

print("\n" + "=" * 75)
print("WHAT DOES NOT MATTER AS MUCH")
print("=" * 75)

print("""
1. User_ID, Post_ID and Comment_ID:
   Technical identifiers, not growth performance metrics.

2. Username:
   Useful for identifying users, but not a direct engagement KPI.

3. Likes alone:
   Evaluate likes together with reach, shares, saves and engagement rate.

4. Total interactions alone:
   Compare average performance per post to avoid volume bias.

5. Gender alone:
   Do not make audience decisions based only on gender.

6. Caption text alone:
   Use caption analysis to understand content, but measure its performance
   using actual engagement metrics.
""")

# =====================================================
# 15. FINAL GROWTH RECOMMENDATIONS
# =====================================================

print("\n" + "=" * 75)
print("5 PRACTICAL INSTAGRAM GROWTH RECOMMENDATIONS")
print("=" * 75)

recommendations = [
    "Focus on the post type with the highest average engagement rate.",
    "Create more content around the topics with the strongest performance.",
    "Prioritize the best-performing posting days and test consistently.",
    "Target cities and account types with stronger engagement and reach.",
    "Reply to comments, understand negative feedback and encourage discussions."
]

for i, recommendation in enumerate(recommendations, 1):
    print(f"{i}. {recommendation}")

# =====================================================
# 16. SAVE ANALYSIS TABLES AS CSV FILES
# =====================================================

for name, table in tables.items():
    filename = name.lower().replace(" ", "_") + ".csv"
    table.to_csv(filename, index=False)

print("\nAll analysis tables saved successfully as CSV files.")
```
